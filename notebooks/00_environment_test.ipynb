```python
import pyodbc

# SQL Server connection settings (Windows Authentication)
server = "localhost"
database = "CustomerChurnDB"
driver = "ODBC Driver 17 for SQL Server"

connection_string = (
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

connection = None
try:
    connection = pyodbc.connect(connection_string, timeout=10)

    # Check the connection without printing machine or account details
    cursor = connection.cursor()
    cursor.execute("SELECT DB_NAME() AS DatabaseName")
    database_name = cursor.fetchone()[0]
    cursor.close()

    print("Successfully connected to SQL Server!")
    print("Database:", database_name)

except pyodbc.Error as error:
    print("SQL Server connection failed:", error)

finally:
    if connection is not None:
        connection.close()
        print("Connection closed.")
```
